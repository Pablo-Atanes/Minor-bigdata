## Onderzoeksvragen

1. Hoe reageert de koers van Microsoft op pieken in de Eurozone en NL inflatiecijfers vergeleken met traditionele "value" aandelen zoals Berkshire Hathaway?
2. Hoeveel maanden vertraging zit er tussen een significante stijging in WTI/Brent olie en een stijging in de CBS Inflatiecijfers in Nederland?
3. Hoe direct is de doorvertaling van de prijzen van Maïs en Tarwe (ZC=F/ZW=F) naar de CPI (Consumentenprijsindex) van het CBS?
4. Wat is de historische correlatie tussen de Aardgas-futures (NG=F) en de NL Werkloosheidscijfers (is er sprake van een vertraagd effect door industriële krimp)?
5. In hoeverre correleert een stijging van de ECB-rente met de waardering van Europese luxegoederen (LVMH), gezien de impact op de kapitaalkosten en consumentenbestedingen?

### Oorzaak, Gevolg en Hypotheses

1. Microsoft (Growth) vs. Berkshire (Value) en Inflatie
- Oorzaak (x): Inflatiecijfers (Eurozone HICP / NL CPI).
- Gevolgen (y): Koersrendement van Microsoft (MSFT) en Berkshire Hathaway (BRK-B).

- H0​: Er is geen significant verschil in de reactie van MSFT en BRK-B op stijgende inflatiecijfers.
- H1​: Bij stijgende inflatie reageert MSFT (growth) negatiever dan BRK-B (value), vanwege de hogere disconteringsvoet van toekomstige kasstromen.

2. Olieprijzen en NL Inflatie (Time Lag)
- Oorzaak (x): Prijsstijgingen van WTI en Brent Olie (CL=F / BZ=F).
- Gevolg (y): NL Inflatiecijfers (CBS).

- H0​: Er zit geen meetbare vertraging (time lag) tussen een stijging in olieprijzen en de stijging in de NL CPI.
- H1​: Er is een significante vertraging van 1 tot 3 maanden voordat een stijging in olieprijzen volledig is doorberekend in de NL CPI.

3. Landbouwgrondstoffen (Maïs/Tarwe) en de CPI
- Oorzaak (x): Prijzen van Maïs (ZC=F) en Tarwe (ZW=F) futures.
- Gevolg (y): Consumentenprijsindex (CPI) van het CBS.

- H0​: Er is geen direct verband tussen de prijzen op de internationale grondstofbeurzen voor Maïs/Tarwe en de lokale CPI in Nederland.
- H1​: Er is een sterke positieve correlatie waarbij stijgende prijzen voor landbouwgrondstoffen leiden tot een stijging in de voedselcomponent van de CPI.

4. Aardgas-futures en Werkloosheid
- Oorzaak (x): Aardgas-futures prijs (NG=F).
- Gevolg (y): NL Werkloosheidscijfer (CBS).

- H0​: Stijgende energieprijzen (aardgas) hebben geen invloed op de werkloosheidscijfers in Nederland.
- H1​: Een aanhoudende stijging van de aardgasprijs leidt met een vertraging tot een hogere werkloosheid door industriële krimp (energie-intensieve sectoren).

5. ECB-rente en LVMH (Luxe goederen)
- Oorzaak (x): ECB Beleidsrente (ECBDFR).
- Gevolg (y): Koers/Waardering van LVMH (MC.PA).

- H0​: Veranderingen in de ECB-beleidsrente hebben geen invloed op de koers van LVMH.
- H1​: Een stijging van de ECB-rente correleert negatief met de koers van LVMH, doordat hogere rentes zowel de consumptie onder druk zetten als de waarderingsmultiples verlagen.
