## 2. Noodzaak van Intensieve Data-Bewerking

Tijdens de dataverkenning liepen we tegen fundamentele problemen aan die de integriteit van ons onderzoek konden bedreigen. We hebben deze als volgt structureel opgelost in onze cleaning service:

### Tijdreeksen Synchroniseren en Gaten Vullen (Weekenden)
Een grote uitdaging met financiële data is dat beurzen in het weekend en tijdens feestdagen dicht zijn. Als we deze dagen simpelweg negeren, lopen de tijdreeksen van verschillende databronnen asynchroon (denk aan crypto, wat wél 24/7 doorgaat). Om alles strak te trekken, dwingen we een kalender-reindex af naar een continue, dagelijkse frequentie (`freq='D'`). 
De lege plekken die zo ontstaan, vullen we veilig op met een **Forward Fill (`ffill`)**. We hebben hier bewust voor gekozen omdat het doortrekken van de laatst bekende slotkoers de realiteit het beste weerspiegelt, zonder ongewenste voorspellingen ('look-ahead bias') te introduceren.

### Strikte Datatypes en Duplicaten Verwijderen
API's zijn onvoorspelbaar. Soms leveren ze op dezelfde dag dubbele waarden of bevatten numerieke velden plotseling tekst. Om falende code verderop te voorkomen:
- Forceren we extreem strikte datatypes. De datums worden gestandaardiseerde Datetime-objecten en prijzen dwingen we numeriek te zijn (`pd.to_numeric(..., errors='coerce')`). Iets wat niet past, wordt een `NaN` en daarna vakkundig weggewerkt door onze *forward fill*.
- Duplicaten op dezelfde datum filteren we er direct uit (`df[~df.index.duplicated(keep='last')]`). We behouden altijd de laatste waarde, omdat eventuele correcties op een handelsdag doorgaans de meest accurate zijn.

### Risico's van Extreme Uitschieters (Outliers) Beperken
Een 'flash-crash' of een simpele typfout in een API kan een extreme piek of dal opleveren. Voor onze statistische modellen zou zo'n uitschieter de berekeningen enorm vertekenen. Om ons hiertegen te beschermen, gebruiken we de **Z-score methode**. 
Wanneer we een waarde tegenkomen die meer dan 3 standaarddeviaties van het gemiddelde ligt, verwijderen we de hele rij niet (want dat verpest onze zorgvuldig gesynchroniseerde kalender weer). In plaats daarvan gebruiken we een techniek genaamd *Winsorizing* (`.clip()`). Hiermee toppen we de extreme waarde letterlijk af op de grens van 3 standaarddeviaties. De structuur blijft intact, maar de verstorende impact is geneutraliseerd.

### De Data Analyse-Klaar Maken
Tot slot ontdekten we dat het vergelijken van absolute waarden onbegonnen werk is (hoe vergelijk je een index van 4000 punten met een aandeel van 15 euro?). Daarom berekent onze cleaning service direct de essentiële afgeleide variabelen, zoals dagelijkse procentuele rendementen (`Returns_pct`) en absolute verschillen (`Diff_abs`). ### 1. Centrale Tendens: Waar ligt het zwaartepunt?

Om een beeld te krijgen van de "normale" waarde van een ticker of indicator, gebruiken we drie verschillende maatstaven. Elke maatstaf belicht een andere kant van de 'gemiddelde' staat van onze portfolio's:

- **Gemiddelde (Mean):** Dit is de rekenkundige optelsom van alle waarden gedeeld door het aantal observaties. In onze analyse gebruiken we dit om de algemene 'koersrichting' of het basisniveau van een macro-economische variabele te bepalen over de gehele tijdsperiode. Het gemiddelde is echter gevoelig voor extreme uitschieters; één dag met een enorme beurscrash trekt het hele gemiddelde naar beneden, ook als de rest van de periode stabiel was.
- **Mediaan (Median):** De mediaan is de middelste waarde wanneer alle data op volgorde van klein naar groot wordt gezet. Het grote voordeel van de mediaan in financiële markten is de robuustheid. Omdat de mediaan niet kijkt naar de *waarde* van de uitersten, maar alleen naar de *positie*, geeft het ons een veel eerlijker beeld van de "typische" handelsdag. Als het gemiddelde veel lager ligt dan de mediaan, weten we direct dat er sprake is van een 'long tail' aan negatieve uitschieters.
- **Modus (Mode):** De modus is de waarde die het meest frequent voorkomt in de dataset. Bij aandelenkoersen die tot op 6 decimalen nauwkeurig zijn, is de modus vaak minder relevant. Echter, voor ons onderzoek naar de **ECB-rente** is de modus cruciaal. Centrale banken laten de rente vaak maandenlang op exact hetzelfde niveau staan. De modus vertelt ons dus wat het 'normale beleid' was tijdens de onderzochte periode, ongeacht tijdelijke verhogingen of verlagingen.

### 2. Spreiding en Risico: Hoe onrustig is de markt?

De centrale tendens vertelt ons waar de data zich bevindt, maar de spreiding vertelt ons hoe betrouwbaar die positie is. In de financiële wereld staat spreiding synoniem voor risico:

### 3. Vorm van de Verdeling: De verborgen risico's

Financiële data volgt zelden een perfecte "normaalverdeling" (de bekende klokvorm). De afwijkingen van deze vorm vertellen ons veel over de verborgen risico's:

- **Skewness (Scheefheid):** 
    - Een **negatieve scheefheid** betekent dat er een langere staart aan de linkerkant zit. In de aandelenmarkt is dit een bekend fenomeen: koersen stijgen vaak gestaag ('climbing the wall of worry') maar kunnen plotseling en hard dalen. Dit vertelt ons dat er een grotere kans is op extreme verliezen dan op extreme winsten.
    - Een **positieve scheefheid** zien we vaak bij grondstoffen. Bij een plotseling tekort aan bijvoorbeeld aardgas of olie kan de prijs in zeer korte tijd exploderen, wat leidt tot een lange staart aan de rechterkant van de verdeling.
- **Kurtosis (Spitsheid):** Dit meet de dikte van de staarten, ook wel 'Fat Tails' genoemd. Een hoge kurtosis betekent dat extreme gebeurtenissen (zwarte zwanen) vaker voorkomen dan je op basis van een theoretisch model zou verwachten. Voor ons onderzoek is dit een cruciale waarschuwing: een hoge kurtosis in de inflatiedata suggereert dat prijsschokken vaak heftiger en grilliger zijn dan de normale marktfluctuaties doen vermoeden.

# Stap 7 — Meervoudige Regressieanalyses

Voor elk van onze 5 onderzoeksvragen voeren we een **meervoudige regressieanalyse** (Multiple Linear Regression) uit. In tegenstelling tot een enkelvoudige regressie (één X-variabele, zoals in de scatter plots van Stap 6) gebruiken we hier voor elk model **meerdere onafhankelijke variabelen** tegelijk. Dit is essentieel omdat economische verbanden zelden in isolatie bestaan: de koers van MSFT wordt niet alléén door Nederlandse inflatie bepaald, maar ook door Eurozone-inflatie en de algemene marktbeweging (waar BRK-B een goede proxy voor is).

## Methodologische keuzes

**Library:** We gebruiken `scikit-learn` (`LinearRegression`) voor het schatten van de coëfficiënten en `scipy.stats` om handmatig p-waarden te berekenen — sklearn levert die namelijk niet standaard, terwijl ze wel essentieel zijn voor de statistische toelichting.

**Frequentie — maandelijks:** Onze macro-economische indicatoren (CBS CPI, CBS Werkloosheid, ECB-rente, Eurozone HICP) worden maandelijks gepubliceerd, terwijl de financiële data dagelijks zijn. Om verschillende frequenties consistent te kunnen vergelijken resamplen we **alle** reeksen naar maandelijkse eindwaarden (`.resample('ME').last()`). Forward-fillen van maand-data naar dagelijks zou ons aantal observaties artificieel opblazen en p-waarden onbetrouwbaar maken.

**Lags:** Voor onderzoeksvragen waar we een vertraagd effect verwachten (RQ2: olie → CPI; RQ4: gas → werkloosheid) nemen we de lag-versies (`.shift(n)`) van de driver expliciet mee als aparte X-variabelen. Zo kan de regressie zelf bepalen welke vertraging het sterkst is.

**Train/test-split:** We splitsen de tijdreeks 80/20 (`shuffle=False`, dus chronologisch — eerste 80% train, laatste 20% test). De RMSE op de test-set geeft een eerlijke maat van out-of-sample voorspelkracht, naast de R² op de train-set.

**Returns vs. niveaus:** Voor aandelenkoersen (RQ1, RQ5) gebruiken we maandelijkse procentuele rendementen als Y in plaats van het prijsniveau. Aandelenprijzen zijn niet-stationair en hebben een trend; rendementen wel. Voor macro-indicatoren in niveaus (CPI, werkloosheid) houden we de niveaus aan, omdat dit conceptueel is wat we willen verklaren.

**Multicollineariteit-noot:** In RQ2 zijn `CL_F` (WTI) en zijn lags onderling sterk gecorreleerd, en evenzo in RQ4 voor `NG_F`. Dit kan individuele coëfficiënten lastig interpreteerbaar maken (grote standaardfouten), maar verandert niets aan het globale model-fit. Lag-modellen zijn gangbaar in tijdreeksanalyse en we vermelden expliciet wanneer dit speelt.

## Hoe lees je de output?

Voor elke RQ tonen we:
1. **Coëfficiëntentabel** — voor elke X-variabele de geschatte coëfficiënt, standaardfout, t-waarde en p-waarde. Een p-waarde < 0.05 (groen gemarkeerd) duidt op statistische significantie.
2. **Modelfit** — R² en adjusted R² (verklaarde variantie), plus train- en test-RMSE.
3. **Scatter werkelijk vs voorspeld** — visuele check van modelfit; punten dicht bij de rode 45°-lijn = goede voorspelling.
4. **Conclusie** — automatisch gegenereerde toelichting per predictor, plus een H0/H1-oordeel.
