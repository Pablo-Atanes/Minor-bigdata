### macro-economische indicatoren
- Inflatie (Consumer Price Index - CPI)
De belangrijkste graadmeter voor de koopkracht. Als dit cijfer stijgt, verhogen centrale banken vaak de rente.
    Yahoo Ticker: CPALTT01USM657N (Consumer Price Index for All Urban Consumers)
    Link: https://finance.yahoo.com/quote/CPALTT01USM657N

- Werkloosheidscijfer (Unemployment Rate)
Een lage werkloosheid duidt op een sterke economie, maar kan ook leiden tot looninflatie.
    Yahoo Ticker: UNRATE
    Link: https://finance.yahoo.com/quote/UNRATE

- Bruto Binnenlands Product (GDP Growth)
De totale waarde van alle goederen en diensten; de ultieme maatstaf voor economische groei.
    Yahoo Ticker: GDP
    Link: https://finance.yahoo.com/quote/GDP

- Fed Funds Rate (De Centrale Rente)
De rente die de centrale bank (de Fed) rekent. Dit bepaalt hoe duur leningen en hypotheken zijn.
    Yahoo Ticker: FEDFUNDS
    Link: https://finance.yahoo.com/quote/FEDFUNDS

- Consumentenvertrouwen (University of Michigan Consumer Sentiment)
Geeft aan hoe optimistisch mensen zijn over hun financiële toekomst. Dit voorspelt vaak toekomstige uitgaven.
    Yahoo Ticker: UMCSENT
    Link: https://finance.yahoo.com/quote/UMCSENT
