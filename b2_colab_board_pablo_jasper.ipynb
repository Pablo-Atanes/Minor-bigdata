## 1. De Workflow in de Notebook

Om te voorkomen dat we met verouderde of handmatig bewerkte bestanden werken, voeren we in ons notebook standaard deze twee stappen uit:

## 2. Noodzaak van Intensieve Data-Bewerking

Tijdens de dataverkenning liepen we tegen fundamentele problemen aan die de integriteit van ons onderzoek konden bedreigen. We hebben deze als volgt structureel opgelost in onze cleaning service:

### Tijdreeksen Synchroniseren en Gaten Vullen (Weekenden)
Een grote uitdaging met financiële data is dat beurzen in het weekend en tijdens feestdagen dicht zijn. Als we deze dagen simpelweg negeren, lopen de tijdreeksen van verschillende databronnen asynchroon (denk aan crypto, wat wél 24/7 doorgaat). Om alles strak te trekken, dwingen we een kalender-reindex af naar een continue, dagelijkse frequentie (`freq='D'`). 
De lege plekken die zo ontstaan, vullen we veilig op met een **Forward Fill (`ffill`)**. We hebben hier bewust voor gekozen omdat het doortrekken van de laatst bekende slotkoers de realiteit het beste weerspiegelt, zonder ongewenste voorspellingen ('look-ahead bias') te introduceren.

### Strikte Datatypes en Duplicaten Verwijderen
API's zijn onvoorspelbaar. Soms leveren ze op dezelfde dag dubbele waarden of bevatten numerieke velden plotseling tekst. Om falende code verderop te voorkomen:
- Forceren we extreem strikte datatypes. De datums worden gestandaardiseerde Datetime-objecten en prijzen dwingen we numeriek te zijn (`pd.to_numeric(..., errors='coerce')`). Iets wat niet past, wordt een `NaN` en daarna vakkundig weggewerkt door onze *forward fill*.
- Duplicaten op dezelfde datum filteren we er direct uit (`df[~df.index.duplicated(keep='last')]`). We behouden altijd de laatste waarde, omdat eventuele correcties op een handelsdag doorgaans de meest accurate zijn.

### Risico's van Extreme Uitschieters (Outliers) Beperken
Een 'flash-crash' of een simpele typfout in een API kan een extreme piek of dal opleveren. Voor onze statistische modellen zou zo'n uitschieter de berekeningen enorm vertekenen. Om ons hiertegen te beschermen, gebruiken we de **Z-score methode**. 
Wanneer we een waarde tegenkomen die meer dan 3 standaarddeviaties van het gemiddelde ligt, verwijderen we de hele rij niet (want dat verpest onze zorgvuldig gesynchroniseerde kalender weer). In plaats daarvan gebruiken we een techniek genaamd *Winsorizing* (`.clip()`). Hiermee toppen we de extreme waarde letterlijk af op de grens van 3 standaarddeviaties. De structuur blijft intact, maar de verstorende impact is geneutraliseerd.
